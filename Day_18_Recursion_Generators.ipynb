```python
def fun(n):
    if n<=0:
        print("Done")
    else:
        print(n,end=",")
        fun(n-1)
fun(10)
```

```python
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
print(fact(5))
```

```python
def fib(n):
    if n <= 1:
        return n
    else:
        return fib(n-1)+fib(n-2)
print(fib(10))
```
