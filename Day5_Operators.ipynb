Identity Operators

```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
```

```python
x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is not y)
```

MemberShipOperators

```python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
```

```python
fruits = ["apple", "banana", "cherry"]

print("pineapple" not in fruits)
```

BitwiseOperators

```python
print(6 & 3)
print(6 | 3)
print(6 ^ 3)
print(~ 3)
print(6 << 3)
print(6 >> 3)
```

OperatorPresidence

```python
print((6 + 3) - (6 + 3))
print(100 + 5 * 3)
```
